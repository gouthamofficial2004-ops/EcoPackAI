```python
# ==========================================
# ECOPACKAI - EXPLORATORY DATA ANALYSIS
# Step 1: Import Libraries and Load Datasets
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned datasets
materials = pd.read_csv("../data/processed/materials_cleaned.csv")
products = pd.read_csv("../data/processed/products_cleaned.csv")
training = pd.read_csv("../data/processed/training_data_cleaned.csv")

print("EcoPackAI - EDA Started Successfully!")

print("\nDataset Shapes:")
print("Materials:", materials.shape)
print("Products:", products.shape)
print("Training Data:", training.shape)
```

```python
# ==========================================
# EDA - STEP 2: DATASET STRUCTURE
# ==========================================

print("MATERIALS DATASET")
display(materials.head())

print("\nPRODUCTS DATASET")
display(products.head())

print("\nTRAINING DATASET")
display(training.head())
```

```python
# EDA STEP 3: STATISTICAL SUMMARY

print("MATERIALS - NUMERICAL SUMMARY")
display(materials.describe())

print("\nPRODUCTS - NUMERICAL SUMMARY")
display(products.describe())

print("\nTRAINING DATA - NUMERICAL SUMMARY")
display(training.describe())
```

```python
# EDA STEP 4: CATEGORICAL ANALYSIS

print("MATERIAL CATEGORY DISTRIBUTION")
display(materials["material_category"].value_counts())

print("\nPRODUCT INDUSTRY DISTRIBUTION")
display(products["industry"].value_counts())

print("\nPRODUCT TYPE DISTRIBUTION")
display(products["product_type"].value_counts())

print("\nSUITABILITY LABEL DISTRIBUTION")
display(training["suitable_label"].value_counts())
```
